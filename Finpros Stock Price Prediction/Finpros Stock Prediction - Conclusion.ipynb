# Tổng kết

+ Model chạy tương đối ổn định với các dataset khác nhau. Nguyên nhân chủ yếu cho sự thiếu chính xác trong model là do tham số chưa được tối ưu ( Nguyên nhân chủ yếu là do thiếu đi tài nguyên tính toán ). Nếu có đủ tài nguyên tính toán, ta có thể tối ưu model bằng cách chạy các hàm tối ưu hóa như GridSearch, RandomSearch, Cross Validation,....

+ Trong quá trình xây dựng model, có một vài khó khăn trong việc tạo ra các hàm để có thể đưa ra dự đoán với time-series và dữ liệu đầu vào có nhiều chiều. Đồng thời việc tìm ra các tham số tối ưu cho model mà không sử dụng các hàm tìm tham số tối ưu mất tương đối nhiều thời gian, tuy nhiên đã có thể tìm ra được một bộ tham số tương đối tối ưu cho model. Model này đã có khả năng dự đoán ổn đến khá ổn, tuy nhiên còn có thể cải thiện độ chính xác hơn nữa.
