```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
```

```python
cols = ["Algebra", "Adv Math", "PsDA", "Geometry", "Ez", "Medium", "Hard"]
df = pd.read_csv(r"C:\Users\phana\PycharmProjects\DataGen\datanew.csv", names=cols)    
#sep = seperator ( any spaces will be recognized as seperator )
```

```python
df.head()
```

```python
a = "Ez"
b = "Medium"
c = "Hard"

X = df[[a,b,c]].values
```

```python
kmeans = KMeans(n_clusters=12).fit(X)
```

```python
labels = kmeans.labels_
```

```python
df['diff%'] = labels
```
